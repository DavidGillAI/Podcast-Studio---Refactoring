```python
"""LLM processor notebook for podcast script generation."""

from pathlib import Path
import sys

ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists()), None)

if ROOT is None:
    raise FileNotFoundError("Could not find the project src directory.")

SRC_DIR = str(ROOT / "src")
if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

class LLMProcessor:
    def __init__(self):
        # Uses OPENAI_API_KEY from your .env file
        self.client = OpenAI()

    def summarize_text(self, text: str) -> str:
        """
        Transform raw transcript text into a 1-minute podcast script.
        """
        prompt = f"""
        You are a podcast script generator.
        You are an educational podcast script writer.
You will receive structured notes extracted from a full class transcript.
Your task is to create a short, engaging recap podcast script for students who missed the lesson.

Requirements
The recap must:

cover the entire lesson, not just the beginning

focus on the main teaching topic

include operational or setup issues only if they affected learning

briefly mention API key or quota issues if they impacted the class

explain the core concepts clearly and simply

sound natural when spoken aloud

be approximately 3–5 minutes when read aloud

Script Structure
Title:  
A short, clear title.

Intro:  
Briefly summarize what happened in class and what the lesson focused on.

Main Lesson Recap:  
Explain the key concepts in a logical, student‑friendly order.

Operational Notes:  
Briefly mention relevant setup/API/project issues only if they mattered for understanding the lesson.

Practical Takeaways:  
List the most important things students should remember or apply.

Outro:  
A short closing line.

        {text}
        """

        # NEW OpenAI Responses API
        response = self.client.responses.create(
            model="gpt-4o-mini",
            input=prompt,
        )

        # Extract the generated text
        return response.output_text
```
