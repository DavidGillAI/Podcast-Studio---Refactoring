```python
"""LLM processor notebook for podcast script generation."""

def log_error(function_name: str, error: Exception, context: str = ""):
    error_message = (
        f"\nERROR in {function_name}\n"
        f"Error Type: {type(error).__name__}\n"
        f"Error Message: {error}\n"
        f"Context: {context}\n"
    )

    print(error_message)

    return error_message

# ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists()), None)

# if ROOT is None:
#     raise FileNotFoundError("Could not find the project src directory.")

# SRC_DIR = str(ROOT / "src")
# if SRC_DIR not in sys.path:
#     sys.path.insert(0, SRC_DIR)

from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

class LLMProcessor:
    def __init__(self):
        try:
            self.client = OpenAI()

        except Exception as e:
            log_error(
                function_name="LLMProcessor.__init__",
                error=e,
                context="Failed to create OpenAI client."
            )
            raise

    def summarize_text(self, text: str) -> str:
        try:
            if not text or not text.strip():
                raise ValueError("No text was provided for summarisation.")

            prompt = f"""You are an educational podcast script writer.

You will receive structured notes extracted from a full class transcript.
Create a short recap podcast script for students who missed the lesson.

The recap must:
- reflect the whole lesson, not only the beginning
- prioritize the actual teaching topic
- include operational context only when it matters
- mention API key / quota issues briefly if they affected the class
- explain the main lesson concepts clearly
- sound natural when read aloud
- be around 3 to 5 minutes when spoken

Use this structure:

Title:
A short title.

Intro:
Briefly explain what happened in class and what the lesson focused on.

Main Lesson Recap:
Explain the main concepts in a clear learning order.

Operational Notes:
Briefly mention relevant setup/API/project issues if important.

Practical Takeaways:
List what students should remember or apply.

Outro:
End with a short closing sentence.

{text}
"""

            response = self.client.responses.create(
                model="gpt-4o-mini",
                input=prompt,
            )

            return response.output_text

        except Exception as e:
            log_error(
                function_name="LLMProcessor.summarize_text",
                error=e,
                context="Failed while generating podcast script with the LLM."
            )
            raise

        # Extract the generated text
        return response.output_text
```

## LLMProcessor Refactoring and Error Handling

The `LLMProcessor` class was refactored to improve modularity, readability, maintainability, and debugging clarity.

### Changes Made

- Added explicit `try/except` handling inside:
  - `__init__()`
  - `summarize_text()`
- Added validation to check whether text input is empty before sending a request to the LLM.
- Integrated the reusable `log_error()` helper function to centralize error reporting.
- Removed unused imports and outdated `src` directory setup code that was no longer relevant to the current project structure.

### Why These Changes Matter

These improvements make the LLM workflow more robust and easier to debug.

The new error handling now clearly shows:
- where an error occurred,
- what type of error occurred,
- the error message,
- and useful contextual information.

This avoids silent failures and makes troubleshooting significantly easier during development.

Removing redundant imports and unused setup code also improves readability and reduces unnecessary technical debt.

### Refactoring Principles Applied

- Modular design
- Separation of concerns
- Reusable helper functions
- Explicit error handling
- Validation before processing
- Removal of redundant/dead code
- “Fail loud, not silently” debugging philosophy
